```python
from pathlib import Path
import numpy as np
import pandas as pd

np.random.seed(42)
num_samples = 500

# Generating baseline synthetic distributions
sleep_hours = np.round(np.random.normal(6.5, 1.5, num_samples).clip(3.5, 9.5), 1)
study_hours = np.round(np.random.normal(6.0, 2.5, num_samples).clip(1.0, 12.0), 1)
academic_workload = np.random.randint(1, 11, num_samples)
caffeine_mg = np.round(
    (study_hours * 35 + np.random.normal(50, 30, num_samples)).clip(0, 600), 0
)
physical_activity = np.round(
    (sleep_hours * 6 + np.random.normal(10, 15, num_samples)).clip(0, 90), 0
)
social_hrs = np.round(
    (sleep_hours * 0.5 + np.random.normal(0, 0.8, num_samples)).clip(0.2, 6.0), 1
)
perceived_control = np.random.randint(1, 11, num_samples)

# Deriving correlated features
heart_rate = np.round(
    70 + (academic_workload * 2.5) - (physical_activity * 0.15) + np.random.normal(0, 3, num_samples)
).astype(int)
gpa = np.round((2.0 + (perceived_control * 0.18) + np.random.normal(0, 0.2, num_samples)).clip(2.0, 4.0), 2)

# Formulating non-linear stress index for target class calculation
stress_score = (
    (10 - sleep_hours) * 1.5
    + (study_hours * 1.2)
    + (academic_workload * 1.8)
    + (caffeine_mg / 100)
    + (heart_rate / 15)
    - (physical_activity / 20)
    - (social_hrs * 1.1)
    - (perceived_control * 1.4)
)

# Binning target variable
def assign_stress_class(score):
    if score < 12:
        return "No Stress"
    elif score < 17:
        return "Low Stress"
    elif score < 22:
        return "Medium Stress"
    else:
        return "High Stress"

stress_classes = [assign_stress_class(s) for s in stress_score]

# DataFrame construction
df = pd.DataFrame({
    "student_id": [f"STU_{i+1:03d}" for i in range(num_samples)],
    "sleep_hours": sleep_hours,
    "study_hours_per_day": study_hours,
    "heart_rate_bpm": heart_rate,
    "academic_workload": academic_workload,
    "caffeine_intake_mg": caffeine_mg,
    "physical_activity_min": physical_activity,
    "social_interaction_hrs": social_hrs,
    "gpa": gpa,
    "perceived_control": perceived_control,
    "stress_level": stress_classes,
})

# Define target directory path and file name
target_dir = Path(r"C:\Users\Rizwan computers\Documents")
file_path = target_dir / "student_stress_dataset_500.csv"

# Ensure the destination folder exists before saving
target_dir.mkdir(parents=True, exist_ok=True)

# Save CSV file
df.to_csv(file_path, index=False)

# Print confirmation message
print(f"Dataset successfully saved to: {file_path}")
```
